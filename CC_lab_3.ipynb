```python
avg_salary = df_emp["Salary"].mean()
total_bonus=df_emp["Bonus"].sum()
youngest_age=df_emp["Age"].min()
highest_rating = df_emp["Rating"].max()
print("Average Salary:", avg_salary)
print("Total Bonus:", total_bonus)
print("Youngest Age:", youngest_age)
print("Highest Rating:", highest_rating)
```

```python
df_sorted=df_emp.sort_values(by="Salary",ascending=False)
df_sorted
```

```python
def categorise_rating(rating):
  if rating>=4.5:
    return "Excellent"
  elif rating >= 4.0:
    return "Good"
  else:
    return "Average"

df_emp["Performance_Category"] = df_emp["Rating"].apply(categorise_rating)
print(df_emp)
```

```python
print(df_emp.isnull().sum())
```

```python
df_emp.rename(columns={"Employee_ID":"ID"},inplace=True)
```

```python
exp_gt_5=df_emp[df_emp["Years_of_Experience"]>5]
print(exp_gt_5)
it_dept=df_emp[df_emp["Department"]== 'IT']
print(it_dept)
```

```python
df_emp["Tax"] = df_emp["Salary"] * 0.10
print(df_emp)
```
