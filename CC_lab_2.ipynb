```python
import random
random.seed(1024160135)
nums =[random.randint(100,900)
for _ in range(100)]
print(nums)
```

```python
odd_count=0
even_count=0
odd_nums=[]
even_nums=[]
for i in nums:
  if(i%2==0):
    even_count+=1
    even_nums.append(i)
  else:
    odd_count+=1
    odd_nums.append(i)
print(f"Odd numbers: {odd_nums}")
print(f"Even numbers: {even_nums}")
print(f"Odd count: {odd_count}")
print(f"Even count: {even_count}")
```

```python
def is_prime(n):
  if n<2:
    return False
  return all(
      n%divisor !=0
      for divisor in range(2,int(n**0.5)+1)
  )

prime_numbers = [num for num in nums if is_prime(num)]
print(prime_numbers)
print(len(prime_numbers))
```

```python
from collections import Counter

frequency = Counter(nums)
most_frequent_number, occurrence_count = frequency.most_common(1)[0]

print("Most frequent number:", most_frequent_number)
print("It occurs", occurrence_count, "times")
```

```python
rollno = int(input("Enter roll no: "))
print(rollno)
a = set()
b = set()
for i in range(0,10):
  x=rollno%10
  rollno=rollno//10
  l=x*7
  a.add(l)
  m=x/7
  m=x*9
  b.add(m)
print(a)
print(b)
```

```python
print(a.union(b))
```

```python
print(a.intersection(b))
```

```python
print(a.difference(b))
print(b.difference(a))
print(a.symmetric_difference(b))
#difference() is directional; A.difference(B) returns elements in A but not in B, while B.difference(A) returns elements in B but not in A; symmetric_difference() returns the combined elements that are in either set but not both.
```

```python
print(a.issubset(b))
print(b.issuperset(a))
```

```python
user_in = int(input("Enter value X:"))
for i in a:
  if(user_in==a):
    a.discard(i)
print(a)
#if the value does not exist, then discard() does nothing but remove() raises a keyerror
```
