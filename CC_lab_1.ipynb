```python
def oddAdd(n):
  sum=0
  while n>0:
    if n%2!=0:
      sum=sum+n
    n=n-1
  return sum
n=int(input("Enter value of n"))
print(oddAdd(n))
```

```python
def primeAdd(n):
  sum=0
  while n>1:
    f=0
    for i in range(2,n//2+1):
      if n%i==0:
        f=1
        break
    if f==0:
      sum=sum+n
    n=n-1
  return sum
n=int(input("Enter value of n"))
print(primeAdd(n))
```
